# Data Segregation

Entrada: `processed.csv`. Saidas: `train.csv` e `test.csv`. A divisao e feita uma unica vez para manter comparabilidade entre os notebooks de treinamento.

```python
from pipeline_utils import load_processed_csv, save_split_csv, split_processed_dataset, write_json

PROCESSED_PATH = "processed.csv"
TRAIN_PATH = "train.csv"
TEST_PATH = "test.csv"
METADATA_PATH = "results/data_segregation_config.json"
TEST_SIZE = 0.2
SEED = 42
```

```python
processed_df = load_processed_csv(PROCESSED_PATH)
train_df, test_df, metadata = split_processed_dataset(
    processed_df,
    test_size=TEST_SIZE,
    seed=SEED,
)

print(f"Training samples: {metadata['n_train']}")
print(f"Test samples: {metadata['n_test']}")
train_df.head()
```

```python
save_split_csv(train_df, TRAIN_PATH)
save_split_csv(test_df, TEST_PATH)
write_json(METADATA_PATH, metadata)

print(f"Saved training dataset: {TRAIN_PATH}")
print(f"Saved test dataset: {TEST_PATH}")
print(f"Saved split metadata: {METADATA_PATH}")
```
