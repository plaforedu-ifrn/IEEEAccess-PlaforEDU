```python
import pandas as pd
import json
from collections import Counter

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

# comp_name_filtered é a coluna com as 53 competências (já filtradas)
# cada célula é uma lista em formato JSON, tipo ["Ensino", "Orientação"]
def parse(cell):
    return json.loads(cell) if isinstance(cell, str) else []

train_counts = Counter()
for lst in train["comp_name_filtered"].apply(parse):
    train_counts.update(lst)

test_counts = Counter()
for lst in test["comp_name_filtered"].apply(parse):
    test_counts.update(lst)

all_labels = sorted(set(train_counts) | set(test_counts))

print(f"Total de competências: {len(all_labels)}")
print(f"Cursos de treino: {len(train)} | Cursos de teste: {len(test)}")
print()
print(f"{'Competência':50s} {'Treino':>7s} {'Teste':>6s}")
print("-" * 66)
for lab in all_labels:
    tr, te = train_counts.get(lab, 0), test_counts.get(lab, 0)
    marca = "  <-- ZERO no teste" if te == 0 else ""
    print(f"{lab:50s} {tr:7d} {te:6d}{marca}")

zero = [l for l in all_labels if test_counts.get(l, 0) == 0]
print()
print(f"Competências com suporte zero no teste: {len(zero)}")
```
