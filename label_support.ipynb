# Label Support Analysis

Reports the per-competency support of the 53 retained competencies across
`train.csv` (270 courses) and `test.csv` (68 courses), identifying competencies
with zero support in the held-out test set. This analysis supports Table 1 of
the paper and the discussion of Macro-F1 handling for zero-support labels.

```python
import pandas as pd
import json
from collections import Counter

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

# comp_name_filtered holds the 53 retained competencies (after rare-label filtering)
# each cell is a JSON-formatted list, e.g. ["Ensino", "Orientação"]
def parse(cell):
    return json.loads(cell) if isinstance(cell, str) else []

train_counts = Counter()
for lst in train["comp_name_filtered"].apply(parse):
    train_counts.update(lst)

test_counts = Counter()
for lst in test["comp_name_filtered"].apply(parse):
    test_counts.update(lst)

all_labels = sorted(set(train_counts) | set(test_counts))

print(f"Total competencies: {len(all_labels)}")
print(f"Training courses: {len(train)} | Test courses: {len(test)}")
print()
print(f"{'Competency':50s} {'Train':>7s} {'Test':>6s}")
print("-" * 66)
for lab in all_labels:
    tr, te = train_counts.get(lab, 0), test_counts.get(lab, 0)
    flag = "  <-- ZERO in test" if te == 0 else ""
    print(f"{lab:50s} {tr:7d} {te:6d}{flag}")

zero_support = [l for l in all_labels if test_counts.get(l, 0) == 0]
print()
print(f"Competencies with zero test support: {len(zero_support)}")
```
