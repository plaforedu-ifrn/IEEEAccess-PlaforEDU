# Processing

Entrada: `dataset.csv`. Saida: `processed.csv`. Este notebook centraliza a limpeza, agregacao por curso e filtro de competencias raras usados pelos modelos supervisionados.

```python
import pandas as pd

from pipeline_utils import process_raw_dataset, save_processed_csv, write_json

RAW_PATH = "dataset.csv"
PROCESSED_PATH = "processed.csv"
METADATA_PATH = "results/processing_config.json"
MIN_COURSE_COUNT = 5
```

```python
raw_df = pd.read_csv(RAW_PATH)
processed_df, metadata = process_raw_dataset(raw_df, min_course_count=MIN_COURSE_COUNT)

print(f"Raw rows: {metadata['n_raw_rows']}")
print(f"Courses before filtering: {metadata['n_courses_before_filtering']}")
print(f"Courses after filtering: {metadata['n_courses_after_filtering']}")
print(f"Labels before filtering: {metadata['n_labels_before_filtering']}")
print(f"Labels after filtering: {metadata['n_labels_after_filtering']}")
processed_df.head()
```

```python
save_processed_csv(processed_df, PROCESSED_PATH)
write_json(METADATA_PATH, metadata)

print(f"Saved processed dataset: {PROCESSED_PATH}")
print(f"Saved processing metadata: {METADATA_PATH}")
```
